```python
# 2. Connect to NanoCam
import serial
ser = serial.Serial('COM3', 115200, timeout=2)
print('Connected!')

def cmd(s):
    ser.write(f'{s}\r\n'.encode())
    import time; time.sleep(0.2)
    return ser.readline().decode().strip()

print(cmd('wifi_ver'))
```

```python
# 3. Get device info
print(cmd('nano_info'))
```

```python
# 4. Switch AI modes
for mode in range(6):
    print(f'Switching to mode {mode}...')
    cmd(f'ai_mode:{mode}')
    import time; time.sleep(3)
```

```python
# 5. View MJPEG stream in OpenCV
import cv2
ip = '192.168.1.100'
cap = cv2.VideoCapture(f'http://{ip}:80/stream')
ret, frame = cap.read()
if ret:
    import matplotlib.pyplot as plt
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.title('NanoCam Live View')
    plt.show()
```
